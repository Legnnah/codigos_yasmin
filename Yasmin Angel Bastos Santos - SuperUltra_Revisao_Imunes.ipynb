### 🎮 Exercício 1: Playlist do Spotify

Objetivo: Trabalhar com listas, métodos e loops

```python
"""
EXPLICAÇÃO:
Vamos criar uma playlist musical! As listas são como suas playlists -
você pode adicionar, remover e organizar músicas.
Dica: Use append() para adicionar e remove() para tirar músicas.
"""

# Sua playlist inicial
playlist = ["Blinding Lights - The Weeknd", "Flowers - Miley Cyrus", "As It Was - Harry Styles"]

print("🎵 SUA PLAYLIST 🎵")
print(playlist)

# 1. Adicione 2 músicas novas que você gosta
# ("Sua música favorita ?")
musica_favorita = input("Qual é a sua música favorita?: ")
# ("Outra música legal ?")
musica_legal = input("Fale outra música que você acha legal: ")
playlist.append(musica_favorita)
playlist.append(musica_legal)

# 2. Remova uma música que enjoou
# Exemplo:
playlist.remove("Flowers - Miley Cyrus")

# 3. Mostre quantas músicas tem na playlist
print(f"Total de músicas: {len(playlist)}")

# 4. Coloque a playlist em ordem alfabética
playlist.sort()

# 5. Mostre a playlist final
print("\n🎵 PLAYLIST ATUALIZADA 🎵")
for musica in playlist:
     print(f"🎧 {musica}")
```

### 📱 Exercício 2: Feed do Instagram
Objetivo: Trabalhar com loops e condicionais

```python
"""
EXPLICAÇÃO:
Vamos simular um feed do Instagram! O 'for' percorre cada post,
e o 'if' decide quais posts você vai curtir.
"""

posts = [
    {"usuario": "Motoboy", "curtidas": 150, "video": False},
    {"usuario": "Pai de familia", "curtidas": 45, "video": True},
    {"usuario": "Virginia", "curtidas": 50000, "video": False},
    {"usuario": "Alejandro", "curtidas": 80, "video": True}
]

print("📱 FEED DO INSTAGRAM 📱")

# 1. Mostre todos os posts
print(posts)
for post in posts:
     print(f"@{post['usuario']} - {post['curtidas']} curtidas")

# 2. Curta apenas os posts com mais de 100 curtidas
print("\n❤️  POSTS QUE VOCÊ CURTIU:")

for post in posts:
    if post['curtidas'] > 100:
        print(f"Você curtiu o post de @{post['usuario']}")


# 3. Mostre apenas os posts que são vídeos
print("\n🎥 VÍDEOS NO FEED:")
for post in posts:
    if post['video'] == True:
        print(f"📹 @{post['usuario']} postou um vídeo!")
```

### 🎯 Exercício 3: Sistema de Pontuação em Jogo

Objetivo: Trabalhar com listas, loops e funções básicas

```python
"""
EXPLICAÇÃO:
Em jogos, usamos listas para guardar pontuações.
Vamos calcular estatísticas dos jogadores!
"""

pontuacoes = [1250, 800, 1560, 920, 2100, 750, 1800]

print("🏆 RANKING DO JOGO 🏆")

# 1. Mostre todas as pontuações
print("Pontuações:", pontuacoes)

# 2. Encontre a maior pontuação (record do jogo)
record = max(pontuacoes)
print(f"🎯 RECORD: {record} pontos!")

# 3. Calcule a pontuação média
total = sum(pontuacoes)
media = total /len(pontuacoes)
print(f"📊 Média: {media:.1f} pontos")

# 4. Mostre apenas as pontuações acima de 1000
print("\n⭐ PONTUAÇÕES DESTAQUE (>1000):")
for pontos in pontuacoes:
     if pontos > 1000:
         print(f"🎮 {pontos} pontos")
```

### 📊 Exercício 4: Análise de Notas da Turma

Objetivo: Trabalhar com tipos de dados e cálculos

```python
"""
EXPLICAÇÃO:
Vamos analisar as notas da turma! Precisamos converter tipos:
- input() sempre retorna texto (string)
- Para calcular, precisamos de números (int/float)
"""

print("📚 ANÁLISE DE NOTAS DA TURMA 📚")

# 1. Coletar notas dos alunos
notas = []
for i in range(5):  # 5 alunos
    nota = float(input(f"Nota do aluno {i+1}: "))
    notas.append(nota)

print(f"\nNotas da turma: {notas}")

# 2. Estatísticas básicas
media = sum(notas) / len(notas)
maior = max(notas)
menor = min(notas)

print(f"\n📊 ESTATÍSTICAS:")
print(f"Média da turma: {media:.1f}")
print(f"Maior nota: {media}")
print(f"Menor nota: {media}")

# 3. Verificar situação da turma
if nota >= 7:
    print("✅ Turma aprovada!")
elif nota >= 5:
    print("⚠️  Turma em recuperação")
else:
    print("❌ Turma reprovada")
```

### 💡 DICAS GERAIS PARA RESOLVER:

**Sobre LISTAS:**

- lista.append(x) - adiciona no final

- lista.remove(x) - remove um item

- len(lista) - conta quantos itens

- max(lista) / min(lista) - maior/menor valor

**Sobre LOOPS:**

- for item in lista: - para cada item

- while condicao: - repete enquanto for verdade

- break - sai do loop

**Sobre CONDICIONAIS:**

- if condicao: - executa se for verdade

- elif outra_condicao: - senão, se outra for verdade

- else: - se nenhuma for verdade

**Sobre TIPOS:**

- int() - converte para número inteiro

- float() - converte para número decimal

- str() - converte para texto

### 🎮 Exercício 1: Perfil de Redes Sociais

Objetivo: Trabalhar com dicionários básicos

```python
"""
EXPLICAÇÃO:
Dicionários são como formulários - têm campos (chaves) e respostas (valores).
Podemos armazenar várias informações sobre uma pessoa em um único dicionário.
"""

print("👤 CRIE SEU PERFIL SOCIAL 👤")

# 1. Crie seu perfil
perfil = {
    "nome": input("Seu nome: "),
    "idade": int(input("Sua idade: ")),
    "cidade": input("Sua cidade: "),
    "seguindo": 0,
    "seguidores": 0
}

# 2. Adicione mais informações
perfil["hobbies"] = input("Seus hobbies (separados por vírgula): ").split(", ")
perfil["musica_favorita"] = input("Sua música favorita: ")

# 3. Mostre o perfil completo
print("\n📋 SEU PERFIL:")
for chave, valor in perfil.items():
        print(f"{'chaves'} : {"valor"}")

# 4. Atualize seguidores
perfil["seguidores"] = 1_500_000
perfil["seguindo"] = 1

print(f"\n📊 ESTATÍSTICAS:")
print(f"Seguidores: {perfil['seguidores']}")
print(f"Seguindo: {perfil['seguindo']}")
```

### 🎯 Exercício 2: Sistema de Jogos Favoritos
Objetivo: Listas de dicionários e filtros

```python
"""
EXPLICAÇÃO:
Uma lista de dicionários é como uma tabela - cada dicionário é uma linha,
e as chaves são as colunas. Perfeito para listar vários itens similares!
"""

jogos_favoritos = [
    {"titulo": "Fortnite", "genero": "Battle Royale", "horas_jogadas": 120, "nota": 8},
    {"titulo": "Minecraft", "genero": "Sandbox", "horas_jogadas": 85, "nota": 10},
    {"titulo": "Valorant", "genero": "FPS", "horas_jogadas": 65, "nota": 9},
    {"titulo": "The Sims", "genero": "Simulação", "horas_jogadas": 45, "nota": 7},
    {"titulo": "FIFA", "genero": "Esporte", "horas_jogadas": 30, "nota": 6}
]

print("🎮 MEUS JOGOS FAVORITOS 🎮")

# 1. Mostre todos os jogos
print("\n📋 TODOS OS JOGOS:")
for jogo in jogos_favoritos:
     print(f"🎯 {jogo['titulo']} - {jogo['genero']}")

# 2. Jogos com mais de 50 horas
print("\n⏰ JOGOS MAIS JOGADOS (>50h):")
for jogo in jogos_favoritos:
     if jogo["horas_jogadas"] > 50:
        print(f"⏱️  {jogo['titulo']}: {jogo['horas_jogadas']}h")

# 3. Encontre o jogo com melhor nota
melhor_jogo = None
for jogo in jogos_favoritos:
     if melhor_jogo is None or jogo["nota"] > melhor_jogo["nota"]:
        melhor_jogo = jogo

print(f"\n🏆 MELHOR JOGO: {melhor_jogo['titulo']} - Nota {melhor_jogo['nota']}")

# 4. Adicione um novo jogo
novo_jogo = {
    "titulo": input("\nAdicione um jogo: "),
    "genero": input("Gênero: "),
    "horas_jogadas": int(input("Horas jogadas: ")),
    "nota": int(input("Nota (0-10): "))
}
jogos_favoritos.append(novo_jogo)
```

### 🏫 Exercício 3: Lista de Amigos da Escola

Objetivo: Dicionários com listas internas

```python
"""
EXPLICAÇÃO:
Dicionários podem conter qualquer tipo de dado, inclusive LISTAS!
Isso permite armazenar múltiplos valores em um único campo.
"""

amigos = [
    {
        "nome": "Yago",
        "turma": "2A",
        "materias_favoritas": ["Matemática", "Artes"],
        "idade": 16,
        "contato": {"email": "ogay@escola.com", "celular": "11-99999-1111"}
    },
    {
        "nome": "Motoboy",
        "turma": "2B",
        "materias_favoritas": ["Educação Física", "História"],
        "idade": 17,
        "contato": {"email": "bruno@escola.com", "celular": "11-99999-2222"}
    },
    {
        "nome": "Leonardo",
        "turma": "2A",
        "materias_favoritas": ["Agricultura", "Pecuária"],
        "idade": 16,
        "contato": {"email": "paidefamilia@escola.com", "celular": "11-99999-3333"}
    }
]

print("👥 LISTA DE AMIGOS DA ESCOLA 👥")

# 1. Mostre informações básicas de todos
print("\n📋 TODOS OS AMIGOS:")
for amigo in amigos:
     print(f"\n👤 {amigo['nome']} - Turma {amigo['turma']}")
     print(f"   📚 Matérias favoritas: {', '.join(amigo['materias_favoritas'])}")

# 2. Amigos da turma 2A
print("\n🎯 AMIGOS DA TURMA 2A:")
for amigo in amigos:
     if amigo["turma"] == "2A":
         print(f"• {amigo['nome']}")

# 3. Contatos de todos
print("\n📞 LISTA DE CONTATOS:")
for amigo in amigos:
     print(f"📱 {amigo['nome']}: {amigo['contato']['celular']}")

# 4. Quem gosta de Matemática?
print("\n🧮 QUEM GOSTA DE MATEMÁTICA?")
for amigo in amigos:
    if "Matemática" in amigo["materias_favoritas"]:
         print(f"➗ {amigo['nome']}")
```

### 💡 DICAS PARA DICIONÁRIOS:

- dicionario["chave"] - acessa um valor

- dicionario.get("chave") - acessa com segurança

- dicionario.keys() - todas as chaves

- dicionario.values() - todos os valores

- dicionario.items() - pares chave-valor

**Modificando:**

- dicionario["nova_chave"] = valor - adiciona/atualiza

- dicionario.update(outro_dict) - atualiza múltiplos valores

- del dicionario["chave"] - remove
