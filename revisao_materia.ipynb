### Looping For

```python
# Conta até 16

for numero in range(17):
    print(numero)

# RANGE
# Significa: Período
```

```python
alunos = ['Nicolas', 'Pedro', 'Thauany', 'Yasmin', 'André', 'Manu', 'Andressa']

for bagre in alunos:
    print(bagre)

# For (Para)
```

### While

```python
while True:
    pergunta = input("Você vai viajar amanhã?")
    if pergunta == "sim":
        print("Boa viagem")
        break


# WHILE (Enquanto)
```

### Exercício 2

```python
while True:
    pergunta = int(input("Digite um número inteiro: "))
    if pergunta == 10:
        print("Correto")
        break

else:
    print("Número errado, tente novamente")


# BREAK (Parar)
```

### Lógica

```python
# Faça um código que peça um número ao usuário
# E depois mostre o número anterior e o próximo
# Exemplo, você digitou o número 8, o anterior é 7 e o sucessor, 9.

pergunta = int(input("Digite um número: "))
anterior = pergunta - 1
sucessor = pergunta + 1
print(f"Ok, o número escolhido foi {pergunta}, portanto o número anterior é {anterior} e o número sucessor é {sucessor}.")
```

```python
# Pedir um número inteiro ao usuário
# Fazer um loop for que vai printar todos os números
# Até o número que o usuário digitou

numero = int(input("Escolha um número inteiro: "))
for numero in range(numero+1):
    print(numero)
```

```python
# Pergunte quantas notas o aluno tem
# Depois pergunte as notas
# Se a nota for maior que 7, print (Boa!)
# Se não, print (Reprovado.)

quantidade_notas = int(input("Quantas notas você tem?: "))

for i in range(quantidade_notas):
    notas = float(input("Quais são as notas?: "))

    if notas >= 7:
        print("Boa! ✨")
    else:
        print("Reprovado ❌")


```
